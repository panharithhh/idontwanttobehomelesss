```python
def get_stats(ids):
    counts = {} 
    
    for pair in zip(ids, ids[1:]):
        counts[pair]  = counts.get(pair, 0) + 1 
    
    return counts 
stats = get_stats(text)    

print(stats)

top_pair = max(stats, key=stats.get)
top_pair
```

```python
def merge(ids,pair,idx): # ids , top pair, new_idx
    
    newids = [] # the new pairs 
    i = 0 
    
    while i < len(ids): 
        if i<len(ids) - 1 and pair[0] == ids[i] and pair[1] ==ids[1+i]:
            newids.append(idx) 
            i += 2 
        else :
            newids.append(ids[i])
            i+= 1  
            
    return newids

new_tokens = 10 

merges = {}
ids = list(tokens)
idx = 256  

for i in range(new_tokens):   
    stats = get_stats(ids)
    top_pair = max(stats, key = stats.get)    
    
    idx += i  
    ids = merge(ids,top_pair,idx)
    merges[top_pair] = idx
    
    
    
```

```python
print(merges)

for (a, b), idx in merges.items():
    print(f"({a}, {b}) -> {idx}")

```

```python
vocab = {idx: bytes(idx) for idx in range(256)}


for (p0,p1),idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids): 
    tokens = b"".join(vocab[idx] for idx in ids)
    tokens = tokens.decode("utf-8", errors="replace")
    return tokens

def enconde(text):
    tokens = text.encode("utf-8")
    
decode("ប្រទេស​")
    
```

```python
tests = [
    "ប្រទេស",                    # country - coeng (្រ subscript)
    "ភ្នំពេញ",                    # Phnom Penh - stacked coeng ្ន
    "សួស្តី",                      # hello - vowel + coeng cluster
    "ខ្ញុំ​ស្រឡាញ់​អ្នក",          # "I love you" WITH ZWSP (​) between words
    "ខ្ញុំស្រឡាញ់អ្នក",            # same, NO ZWSP - compare token counts
    "កម្ពុជា 2025 🇰🇭",          # Khmer + ASCII digits + emoji flag
    "១២៣៤៥",                      # Khmer digits (≠ ASCII 12345)
    "​",                          # lone ZWSP (U+200B) - easy to silently drop
    "៚",                          # rare Khmer sign
    "",                           # empty string
]

for s in tests:
    enc = encode(s)
    ok = decode(enc) == s
    print(f"{ok}  len={len(enc):3d}  {repr(s)}")
```

```python
from minbpe import BasicTokenizer

tok = BasicTokenizer()
tok.load("khmer_basic.model")          # loads merges + rebuilds vocab

# --- 1) show learned subword tokens (skip partial-byte ones) ---
print("learned Khmer tokens:")
for idx in range(256, len(tok.vocab)):
    piece = tok.vocab[idx].decode("utf-8", errors="replace")
    if "\ufffd" not in piece:           # hide incomplete-UTF8 fragments
        print(f"{idx:>4}  {piece!r}")

# --- 2) encode a sentence and see the actual token pieces ---
sample = "ប្រទេសកម្ពុជាមានទីក្រុងភ្នំពេញ"
ids = tok.encode(sample)
pieces = [tok.vocab[i].decode("utf-8", errors="replace") for i in ids]

print(f"\nsample : {sample}")
print(f"tokens : {len(ids)}  (from {len(sample.encode('utf-8'))} utf-8 bytes)")
for i, p in zip(ids, pieces):
    print(f"  {i:>4}  {p!r}")
```
